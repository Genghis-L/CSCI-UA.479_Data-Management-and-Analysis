## Relational databases

Relational databases organize data in a collection of tables (relations). You can probably describe the characteristics of a relational database because we've already worked with tabular data.

- each table has named columns, with the actual data that populates the table in separate rows

- rows are sometimes referred to as records or tuples

- each table represents a kind or type of record or tuple, with every row representing an instance of that kind of record

Examples of Relational databases

- MySQL / MariaDB (the latter is a fork of the former)
- PostgreSQL
- Oracle
- Microsoft SQL Server

## SQLite

SQLite supports five basic data types to store its column values:
- NULL: for representing missing data
- INTEGER: uses a variable amount of bytes depending on the value's magnitude
- REAL: stores a floating point value in eight bytes
- TEXT: a text string
- BLOB: A "blob" of data (binary large object) which can be anything, stored just AS it wAS input. Can be used to store images, for instance.

### SELECT

The SELECT command executes in the following order:
1. FROM: SELECTs the table(s)
2. WHERE (if any) : applies conditions
3. GROUP BY (if any): creates groups
4. Aggregations (if any): aggregates by groups, typically using a function LIKE avg or max
5. HAVING (if any): conditions on the aggregated results
6. SELECT: SELECT columns
7. Distinct (if any): apply this if specified
8. ORDER BY (if any): apply order if specified
9. Limit (limit the number of records output)

In particular, WHERE is used for filtering results: 
- conditions can use operators LIKE $=,\langle \rangle$ (not equal), $>,<$
- you can also use LIKE with \% representing wildcards to match on substrings (cASe-insensitive)
- use col_name IS NULL to check for a NULL value
- multiple conditions can be put together with AND, OR, NOT
- parentheses can be ADDed to specify precedence

```python
-- The bASic structure of a SELECT comamnd in SQL is AS follows:
SELECT [DISTINCT] columns
  FROM table
  WHERE condition
  GROUP BY columns
  HAVING condtion
  ORDER BY column [ASC | DESC]
  LIMIT offSET, count
-- SELECT and FROM are required; all of the rest are optional.
```

```python
print(pd.read_sql_query("SELECT * FROM movie limit 10", conn))
```

```python
# SELECT just the title, director and genre WHERE the genre contains the string "Thriller" someWHERE in it.
query="""
SELECT title, director, genre 
FROM movie 
WHERE genre LIKE '%Thriller'
"""
print(pd.read_sql_query(query, conn))
```

```python
# SELECT just the title, director and genre WHERE the genre is not "Science Fiction"
query="""
SELECT title, director, genre 
FROM movie 
WHERE genre<>'Science Fiction'
"""
print(pd.read_sql_query(query, conn))
```

```python
# SELECT and remove duplicated directors by DISTINCT
query="""
SELECT DISTINCT director 
FROM movie
"""
print(pd.read_sql_query(query, conn))
```

- SELECT by sorting

```python
query="""
SELECT title, director, budget 
FROM movie 
ORDER BY director ASC, budget DESC
"""
print(pd.read_sql_query(query, conn))
```

- SELECT by grouping and aggregating

```python
# use GROUP BY in conjunction with COUNT and ORDER BY to find out how many movies each director directed.
print(pd.read_sql_query("SELECT director, count(*) FROM movie GROUP BY director ORDER BY director", conn))
```

```python
# using `HAVING` instead of WHERE for the aggregated feature
query="""
SELECT director, avg(budget) AS avg_budget
FROM movie
WHERE runtime>125
GROUP BY director
HAVING avg_budget>20000000
ORDER BY director ASC
"""
print(pd.read_sql_query(query, conn))
```

```python
# Group Concatenation
query="""
SELECT director, group_concat(title) AS movie_list FROM movie
GROUP BY director ORDER BY director
"""
print(pd.read_sql_query(query, conn))
```
